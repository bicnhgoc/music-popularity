### **II. Data Preprocessing**

```python
data = pd.read_csv('billboard_hot_100_spotify_audio_features_enriched.csv')
```

```python
data.info()
```

```python
#exclude irrelevant columns
data = data.drop(columns=['spotify_track_preview_url','spotify_track_id','spotify_track_album','evidence_span','cot_rationale'])
```

The reason why I excluded these columns is not only because they are not relevant to the analysis, but also due to purpose-overlapped to other columns (*'song_id'* and *'spotify_track_id'*) and large amounts of null variables (e.g. *'spotify_track_preview_url'* has 48% of null records). 

```python
#drop null values
data = data.dropna()
```

```python
#convert data types
data['spotify_track_duration_ms'] = data['spotify_track_duration_ms'].astype(int)
data['spotify_track_explicit'] = data['spotify_track_explicit'].astype(bool)
data['first_week'] = pd.to_datetime(data['first_week'], format='%Y-%m-%d')
data['last_week'] = pd.to_datetime(data['last_week'], format='%Y-%m-%d')
```

```python
#for the music genre i want to convert the lists into strings so that they can be used in the model.
data['spotify_genre'] = data['spotify_genre'].apply(lambda x: ', '.join(eval(x)) if isinstance(x, str) else x)    
```
